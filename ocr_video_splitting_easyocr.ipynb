```python
import cv2
import easyocr

# Initialize EasyOcr reader
reader = easyocr.Reader(['en'], gpu=True)

video_path = '/content/drive/MyDrive/OCR_images/kids_books_letters.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print('Video could not be opened')
    exit()

fps=int(cap.get(cv2.CAP_PROP_FPS))
print(f'Number of Video FPS: {fps}')

frame_count = 0
extracted_texts = set() # Used a tuple instead of a list to delete duplicates

while cap.isOpened():
    return_status, frame =cap.read()

    if not return_status:
        break

    if frame_count % fps == 0: # Skips none zero remainer to save memory, multiply by 2 for a faster response
        current_second = frame_count // 2

        results = reader.readtext(frame, detail=0) # detail=0 returns only the raw text instead of coordinates

        if results:
            print(f"In second {current_second}, found text:{results}")

            for text in results:
                extracted_texts.add(text.strip())

    frame_count+=1

cap.release()
cv2.destroyAllWindows()

print("found texts...")
for i in sorted(extracted_texts):
    print(f"* {i}")
```
